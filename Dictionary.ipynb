### Sort Dictionary by Value

```python
import operator
#operator module provides function for common operations like sorting or itemgetter
d={12:4,3:9,2:1,10:5} # creating a dictionary
print("original dict",d) #printing the original dictionary
sorted_d=sorted(d.items(),key=operator.itemgetter(1)) #sorting the items based on values (first element of each pair)
print("dictionary in ascending order by value: ",sorted_d) #printing the dict in ascending order
#reversing the string
sorted_d=dic(sorted(d.items(), key=operator.itemgetter(1), reverse=True))
print("dictionary in reverse",sorted_d) #printing the dict in reverse order
```

### Concatenate Dictionaries

```python
north_sales = {
    'January': 12,
    'February': 15,
    'March': 10
}
south_sales = {
    'January': 8,
    'February': 11,
    'March': 14
}
east_sales = {
    'January': 10,
    'February': 9,
    'March': 13
}
west_sales = {
    'January': 14,
    'February': 13,
    'March': 9
}
#Method 1st for combining dictionary
region_sales={**east_sales,**west_sales,**north_sales,**south_sales}
print(region_sales)
#Method 2nd for combining dictionary
region_sales={}
region_sales.update(east_sales)
region_sales.update(west_sales)
region_sales.update(north_sales)
region_sales.update(south_sales)
print("region_sales :",region_sales)
```

### Check Key Existence in Dictionary

```python
#Check if "headphones" exist in "Electronics" category, and if available stock is greater than 0.
#If not available, print "Out of Stock" else print "In Stock".
product_stock = {
    "Electronics": {"laptop": 12, "headphones": 0, "camera": 5},
    "Groceries": {"rice": 30, "milk": 0, "bread": 15},
    "Books": {"python_guide": 7, "novel": 0}
}
def is_present(category,item):
    category = category.capitalize()
    if category in product_stock and item in product_stock[category]:
        if product_stock[category][item]>0:
            print("In Stock")
        else: print("Out of stock")
    else: print("Item or Category not found in product_stock")
is_present("Electronics","laptop")
is_present("Daily use","milk")
is_present("Books","python_guide")



```

### Get Maximum and Minimum Values of a Dictionary

```python
gdp_data = {
    "USA": 26.7, "China": 17.9, "Japan": 4.2, "Germany": 4.0,
    "India": 3.7, "UK": 3.2, "France": 3.1, "Brazil": 2.1,
    "Canada": 2.2, "Russia": 1.9}
max_gdp=max(gdp_data,key=gdp_data.get)
min_gdp=min(gdp_data,key=gdp_data.get)
print(max_gdp)
print(min_gdp)
#max\min(gdp_data) compares on the value of keys but key,gdp_data will let compare the keys based on values and return (max and min ) keys with highest and lowest values
```
