```python
#days between dates
from datetime import datetime
print("enter date in dd-mm-yyyy format ")
d1=input()
d2=input()
d1_obj=datetime.strptime(d1,"%d-%m-%Y")
d2_obj=datetime.strptime(d2,"%d-%m-%Y")
diff=abs(d2_obj-d1_obj)
print(diff,"days")
```
