### Count character frequency in a string

```python
def char_frequency(str1):
    dict={}
    for  n in str1:
        keys=dict.keys()
        if n in keys:
            dict[n]+=1
        else:
            dict[n]=1
    return dict
print(char_frequency("shiva upadhyay"))
```

### Replace first char occurrences with $.

```python
input_str=input("enter your string")
char1=str(input_str[0])#accessing first character
if char1 in input_str[1:]: # starting loop from second character
    new_string=char1+input_str[1:].replace(char1,"$")#added first character and started loop from first character
print(new_string)

#things to remember
# dont use len =len("") it overrides len functions
# replace function store variable in another variable
```

### Replace 'not'...'poor' with 'good'

```python
sam_string=input("enter the string")
if "not" in sam_string and "poor" in sam_string:
    start=sam_string.find("not")
    end=sam_string.find("poor")
    if start<end:
        remove=sam_string[start:end+4]
        new_string1=sam_string.replace(remove,"good")
        print("replaced string :",new_string1)
else: print(sam_string)


```

###  Find longest word in a list

```python
lstring=input("enter the string with space")
max=0
lstring=lstring.split(" ")
for char in lstring:
    length=len(char)
    if length>max:
        max=length
        longest_word=char

print(longest_word)
print(max)

```

### Reverse string if length is a multiple of 4

```python
rev_str=input("enter the string")
if len(rev_str)%4==0:
    reversed_string=rev_str[::-1]
    print("reversed string : ",reversed_string)
else: print(rev_str)
```
