```python
#cut() – Fixed-Range Binning
# Bucket age into fixed ranges
bins=[0,25,45,df["age"].max()]
labels=["Youth","Young Adults","Seniors"]
df["age_group"]=pd.cut(df["age"],bins=bins,labels=labels)
df[["age","age_group"]].head(5)
```

```python
# Bucket people based on hours-per-week
bins=[0,25,75,99]
labels=["low","normal","high"]
df["working_hours"]=pd.cut(df["hours-per-week"],bins=bins,labels=labels)
df.head(5)
```

```python
#Topic: qcut() – Quantile-Based Binning
#Bucket people based on hours-per-week into 4 quartiles
df["workload_quartile"] = pd.qcut(df["hours-per-week"], q=5, labels=["Q1", "Q2", "Q3", "Q4"],duplicates="drop") #remove duplicates edges
```

```python
"""to categorize capital-gain into:
No Gain (0)
Low Gain (1–5000)
High Gain (>5000)
"""
bins=[-1,0,5000,df["capital-gain"].max()]
labels=["No Gain","Low Gain","High Gain"]
df["capital_cat"]=pd.cut(df["capital-gain"],bins=bins,labels=labels,duplicates="drop")
```
