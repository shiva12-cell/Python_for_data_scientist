#### Remove occurence of particular item from a list 

```python
b=[2,3,4,5,6,7,8,9,1,2,3,5,6,3,3,4,5,6,5,3,3]
num=int(input("Enter num to delete its occurence in list"))
for i in b:
    if num==i:
        b.remove(i)
print("removing element",num)
print(b)
```

#### Check if we have common element in  2 list

```python
l1=[1,2,3,4]
l2=[3,4,5,6]
for i in l1:
    for j in l2:
        if i==j:
            print(i)
            break
```

```python
# Removing duplicate value from the list 
l2=[1,2,3,4,1,2]
for i in l2:
    count=0
    for j in l2:
        if i==j:
            count=count+1
            if count>1:
                l2.remove(j)
print(l2)
```

### Write a Python program to map two list into a dictionary 

```python
Name=["Shiva","Nandini","Rudra","Sumit"]
Age=[22,28,15,20]
d2={} # declare an empty dictionary
l=len(Name)
j=0
for i in Name:
    d2[i]=Age[j]
    j=j+1
print(d2)
```

#### WAP to sort the dict by keys

```python
d3= {
    22:"Shiva",
    28:"Nandini",
    29:"Ashish",
    21:"Sumit",
    15:"Rudra"
}
a=sorted(d3,reverse=True)
d4={} #intialising a new dictionary

for i in a:
    d4[i]=d3[i]
print(d4)
```

#### WAP to get the top three city by temperature which are stored in a dictionary

```python
temp = {
    "Delhi": 38.5,
    "Mumbai": 32.0,
    "Bangalore": 28.4,
    "Chennai": 35.2,
    "Kolkata": 34.0,
    "Shimla": 18.5,
    "Jaipur": 40.1,
    "Hyderabad": 31.8
}
val=sorted(temp.values(),reverse=True)
print(val)

for i in range(0,3):  # this will iterate thru sorted list 
    for j in temp:
        if val[i]==temp[j]:
            print(j)
```

#### Check how many times a word is there in sentence
string split values loop dictionary (interview questions)

```python
sentence = "apple banana apple cherry apple banana cherry orange apple"
b=sentence.split() # get split in list 
d5={}
for i in b:   # start loop to check each individual element
    if d5.get(i) == None: # if element is first time occuring
        d5[i]=1  # store its value 1
    else:
        d5[i]=d5[i]+1 # if element is occuring again : increase the value by  1 and store in dictionary 
print(d5)


```
