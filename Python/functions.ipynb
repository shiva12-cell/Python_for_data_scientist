### Write a function to count the number of uppercase character in string

```python
def upper_count(name):# here name : parameter
    count=0
    for i in name:
        if i.isupper():
            count=count+1
    return count
    
```

```python
upper_count("Shiva Upadhyay JECRC University")  # my name is argument
```

### Write a calculator function

```python
def calc(a,b,c):
    ''' This is the documentation for the function 
    a is the first number
    b is the second number
    c is the operation we want to perform.
    '''
    if c== "+":
        return a+b
    elif c=="-":
        return a-b
    elif c=="*":
        return a*b
    elif c=="/":
        return a/b
solution=calc(2,3,"-")
solution2=calc(4,7,"*")
print(solution,solution2)        
```

```python
calc() # shift + tab for documentation
```

```python
# user input 
num_1=int(input("Enter the first number")) #  the first number
num_2=int(input("Enter the second number")) #  the second number 
operator=input("Enter the operation u want to perform")
result = calc(num_1,num_2, operator) # calling the function
print(result) #printing the result
```

### Flexible Argument Assignment

```python
def month_salary(hrs,weeks,pay_per_hours):
    salary=hrs*weeks*pay_per_hours
    return salary
```

```python
salary=month_salary(40,4,500)
print(salary)
# For flexible arguement 
# we should pass arguments with parameter name 
salary1=month_salary(hrs=40,pay_per_hours=600,weeks=4)
print(salary1)
```

### Variable number of arguments  
#### When we dont know number of arguments are passed 
#### (use *_____) *parametername 




```python
# average of marks ( we dont know num of parameter)
def avg_grade(*marks):
    total,count=0,0
    for i in marks:
        total =total+i
        count = count +1
    avg=float(total/count)
    print(avg)
avg_grade(20,30,40,55,65,78,81,29)


    
```
