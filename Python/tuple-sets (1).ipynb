### Check whether a string is a palangram or not 
#### Palangram is a sentence which contain every letter in english alphabets.
Eg. The quick brown fox jumps over the lazy dog.

```python
a= "The quick brown fox jumps over the lazy dog"
a=a.lower()
b=set(a)
count=0
for i in b:
    if i.isalpha()==True:
        count=count+1
if count==26:
    print("The string is palangram")
else :
    print("The string is not a palangram number")

```

### WAP to check if B set is a subset or not of A

```python
A={3,4,5,6}
B={3,4,5}
count=0
for i in B:
    for j in A:
        if i==j:
            count=count+1
            break
if count==len(B):
    print("B is subset of A")
else:
    print("B is not Subset of A")
```
