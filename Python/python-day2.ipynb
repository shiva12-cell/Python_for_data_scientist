### Find the 20 odd number (when we dont know the range)


```python
i=1
count=0
while True:
    if i%2==1: # Checking if the value of 1 is odd or not
        count = count+1 # everytime we encounter an odd number, counter will increase
    if count==20:
        print(i)
        break
    i=i+1 
print("Loop is executed")
```

### Continue Statement
It skip/rejects all the code in current iteration of the loop and moves control to top of the loop

```python
i=0
while i<=5:
    i=i+1
    if i==3:
        continue # skips 
    print(i)
```

```python
a="Data Analytics"
for i in a:
    if i=="a":
        continue
    print(i)
```
