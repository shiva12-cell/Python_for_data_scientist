# 006 - Write a function
 * You are given the year, and you have to write a function to check if the year is leap or not.
 * Output : Boolean

```python
def is_leap(year):
    if year%4==0:
        if year%100==0 and year%400==0:
            return "leap"
        elif year%100==0:
            return "not_leap"
        else : return "leap"
    else : return "Not_leap"

year = int(input())
print(is_leap(year))
```
