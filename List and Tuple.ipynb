###  Check if an Array Contains Any Duplicate Elements

```python
def is_duplicate(num_array):
    set_array=set(num_array)
    if len(num_array)==len(set_array):
        return True
    else: return False
print(is_duplicate([1,2,3,4,5,6]))
print(is_duplicate([1,2,2,3,4,5]))
#concept len(array) and len(string) will be equal for unique array otherwise less then len(array)
```

### find first duplicate in the given array

```python
def first_duplicate(num_array):
    c=0
    num_set=set(num_array)
    if len(num_array)==len(num_set):
        print("No duplicates")
    else :
        for i in num_set:
            if i in num_array:
                c+=1
            if c==2:
                print(i)
first_duplicate([1,2,2,3,3,4,5])
first_duplicate([1,2,3,4,5,6])



```
